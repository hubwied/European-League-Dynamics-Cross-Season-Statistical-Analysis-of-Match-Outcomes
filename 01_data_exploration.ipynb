```python
import pandas as pd
```

```python
url_prl = "https://www.football-data.co.uk/mmz4281/2526/E0.csv"
df = pd.read_csv(url_prl)
```

```python
missing_values = df.isnull().mean() *100
print("percentage of missing values:")
print(missing_values.sort_values(ascending=True).head(50))
```

```python
my_features = [
    'Date', 'Time', 'HomeTeam', 'AwayTeam', 
    'FTHG', 'FTAG', 'FTR', 
    'HTHG', 'HTAG', 'HTR',
    'HS', 'AS', 'HST', 'AST', 
    'HR', 'AR', 
    'B365H', 'B365D', 'B365A'
]
```

```python
df_clean = df[my_features].copy()
```

```python
rename_mapping = {
    'FTHG': 'home_goals',
    'FTAG': 'away_goals',
    'FTR': 'match_result',
    'HS': 'home_shots',
    'AS': 'away_shots',
    'HST': 'home_shots_target',
    'AST': 'away_shots_target',
    'HR': 'home_red_cards',
    'AR': 'away_red_cards'
}
```

```python
df_clean.rename(columns=rename_mapping, inplace=True)
```

```python
display(df_clean.head())
```

```python
df_clean.to_csv('01_clean_data.csv', index=False)
```

```python
missing_values_clean = df_clean.isnull().mean() *100
print("percentage of missing values in cleaned dataset:")
print(missing_values_clean)
```
